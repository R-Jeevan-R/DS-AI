```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
y_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor = X_train_tensor.to(dtype = torch.float32)
```

```python
y_train_tensor = y_train_tensor.to(dtype = torch.float32)
```

```python
X_test_tensor = X_test_tensor.to(dtype = torch.float32)
```

```python
y_test_tensor = y_test_tensor.to(dtype = torch.float32)
```

```python
class MySimpleNN():

  def __init__(self, X):
    self.weights = torch.rand(X.shape[1], 1, dtype = torch.float32, requires_grad = True)
    self.bias = torch.zeros(1, dtype = torch.float32, requires_grad  = True)

  def forward(self,X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1- epsilon)

    loss = - (y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
learning_rate = 0.1
epochs = 25
```

```python
# Create Model
model = MySimpleNN(X_train_tensor)

# define loops

for epoch in range(epochs):

  # Forward Pass
  y_pred = model.forward(X_train_tensor)

  # Calculate Loss
  loss = model.loss_function(y_pred, y_train_tensor)

  print(f"Epoch : {epoch + 1}, Loss: {loss.item()}")

  # Backward pass

  loss.backward()

  # Parameter Updates
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad
  model.weights.grad.zero_()
  model.bias.grad.zero_()
```

```python
# Model Evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.5).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f"Accuracy: {accuracy.item()}")
```

## Sequential Container

```python
class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out



```

```python
# Create Dataset

features = torch.rand(10,5)

model = Model(features.shape[1])
model(features)

```

## Improving Our Original Task


```python
class MySimpleNN(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)
    return out

  def loss_function(self, y_pred, y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1- epsilon)

    loss = - (y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
learning_rate = 0.1
epochs = 25
```

```python
# Create Model
model = MySimpleNN(X_train_tensor.shape[1])

# define loops

for epoch in range(epochs):

  # Forward Pass
  y_pred = model(X_train_tensor)

  # Calculate Loss
  loss = model.loss_function(y_pred, y_train_tensor)

  print(f"Epoch : {epoch + 1}, Loss: {loss.item()}")

  # Backward pass

  loss.backward()

  # Parameter Updates
  with torch.no_grad():
    model.network.linear.weight -= learning_rate * model.network.linear.weight.grad
    model.network.linear.bias -= learning_rate * model.network.linear.bias.grad
  model.network.linear.weight.grad.zero_()
  model.network.linear.bias.grad.zero_()
```

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor = X_train_tensor.to(dtype = torch.float32)
```

```python
y_train_tensor = y_train_tensor.to(dtype = torch.float32)
```

```python
X_test_tensor = X_test_tensor.to(dtype = torch.float32)
```

```python
y_test_tensor = y_test_tensor.to(dtype = torch.float32)
```

```python
class MySimpleNN():

  def __init__(self, X):
    self.weights = torch.rand(X.shape[1], 1, dtype = torch.float32, requires_grad = True)
    self.bias = torch.zeros(1, dtype = torch.float32, requires_grad  = True)

  def forward(self,X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1- epsilon)

    loss = - (y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
learning_rate = 0.1
epochs = 25
```

```python
# Create Model
model = MySimpleNN(X_train_tensor)

# define loops

for epoch in range(epochs):

  # Forward Pass
  y_pred = model.forward(X_train_tensor)

  # Calculate Loss
  loss = model.loss_function(y_pred, y_train_tensor)

  print(f"Epoch : {epoch + 1}, Loss: {loss.item()}")

  # Backward pass

  loss.backward()

  # Parameter Updates
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad
  model.weights.grad.zero_()
  model.bias.grad.zero_()
```

```python
class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
learning_rate = 0.1
epochs = 25
```

```python
loss_function = nn.BCELoss()
```

### Batch Gradient Descent

```python
# Create Model
model = Model(X_train_tensor.shape[1])

# define loops
# Define optimizer

optimizer = torch.optim.SGD(model.parameters(), lr = learning_rate)

#Batch Gradient Descent -- Slower Convergence
for epoch in range(epochs):

  # Forward Pass
  y_pred = model(X_train_tensor)

  # Calculate Loss
  loss = loss_function(y_pred, y_train_tensor.view(-1,1))

  # Clear Gradients
  optimizer.zero_grad()

  # Backward pass

  loss.backward()

  # Parameter Update
  optimizer.step()

  print(f"Epoch : {epoch + 1}, Loss: {loss.item()}")
```

### Mini Batch GD

```python
# Mini-Batch GD

batch_size = 32
epochs = 25
n_samples = len(X_train_tensor)

for epoch in range(epochs):
  #Loop over the dataset in chunks of "batch_size"
  for start_idx in range(0, n_samples, batch_size):
    end_idx = start_idx + batch_size
    X_batch = X_train_tensor[start_idx:end_idx]
    y_batch = y_train_tensor[start_idx:end_idx]

    y_pred = model(X_batch)
    loss = loss_function(y_pred, y_batch.view(-1,1))
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

  print(f"Epoch : {epoch + 1}, Loss: {loss.item()}")
```

### Read Dataset and DataLoader notebook and then proceed

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):

  def __init__(self,features, labels):
    self.features = features
    self.labels = labels

  def __len__(self):
    return self.features.shape[0]

  # We can make any transformations here in this method before returning item
  def __getitem__(self, index):
    return self.features[index], self.labels[index]
```

```python
train_dataset = CustomDataset(X_train_tensor, y_train_tensor)
test_dataset = CustomDataset(X_test_tensor, y_test_tensor)
```

```python
train_dataset[3]
```

```python
train_loader = DataLoader(train_dataset, batch_size = 32, shuffle = True)
test_loader = DataLoader(test_dataset, batch_size = 32, shuffle = True)
```

```python
class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

## Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

```python
# Creating model
model = Model(X_train_tensor.shape[1])

# Define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr = learning_rate)

# Define Loss Function
loss_funciton = nn.BCELoss()
```

## Training pipeline

```python
for epoch in range(epochs):
  for batch_features, batch_labels in train_loader:
    y_pred = model(batch_features)
    loss = loss_function(y_pred,batch_labels.view(-1,1))
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
  print(f"Epoch : {epoch + 1}, Loss: {loss.item()}")
```

## Evaluation

```python
model.eval()
accuracy_list = []
with torch.no_grad():
  for batch_features, batch_labels in test_loader:
    print("Shape of bacth labels : ",batch_labels.shape)
    y_pred = model(batch_features)
    print("Shape of predicted labels : ", y_pred.shape)
    y_pred = (y_pred > 0.8).float()
    batch_accuracy = (y_pred.view(-1) == batch_labels).float().mean().item()
    accuracy_list.append(batch_accuracy)

overall_accuracy = sum(accuracy_list) / len(accuracy_list)
print(f'Accuracy : {overall_accuracy:.4f}')
```

## Example 3 :

1.   z = w.x + b
2.   y_pred = sigmoid(z) = 1 / (1 + e^(-z))
3.   loss = -[y_target * ln(y_pred) + (1 - y_target) * ln(1 - y_pred)]






```python
x = torch.tensor([2., 3., 4.], requires_grad=True)
w = torch.tensor([1.,1.1,2.], requires_grad=True)
b = torch.tensor(1. , requires_grad=True)
y_target = torch.tensor(0.)
```

```python
z = torch.mul(x,w) + b
```

```python
z
```

```python
def sigmoid(x):
  return 1 / (1 + torch.exp(-x))
```

```python
y_pred = torch.sigmoid(z)
```

```python
y_pred
```

```python
def Loss(y_hat, y):
  return -(y * torch.log(y_hat) + (1 - y) * torch.log(1 - y_hat))
```

```python
L = Loss(torch.max(y_pred), y_target)
```

```python
L
```

```python
L.backward()
```

```python
x.grad
```

```python
w.grad
```

```python
b.grad
```

## Example 4 :

-- Clearing gradients

```python
x = torch.tensor([2., 4., 6., 8.],requires_grad=True)
```

```python
x
```

```python
for _ in range(3):
  y = torch.mean(torch.square(x))
  y.backward()
  print(x.grad)
```

```python
x.grad.zero_()
```

#### Gradients get accumulated with each forward pass or iteration, which is not desirable and we should manually set them to zero after each iteration

```python
for _ in range(3):
  y = torch.mean(torch.square(x))
  y.backward()
  print(x.grad)
  x.grad.zero_()
```

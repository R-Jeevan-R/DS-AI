##### Normal Distribution - Normal distribution, also known as Gaussian distribution, is a probability distribution that is commonly used in statistical analysis. It is a continuous probability distribution that is symmetrical around the mean, with a bell-shaped curve.
![alt text](Probability-Density-for-Normal-Distribution.png)

```python
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
def normal(m, s):
    return np.random.normal(m, s, size=1000)
```

```python
sns.kdeplot(normal(0, 1), label='N(0, 1)', fill=True)
sns.kdeplot(normal(0, 2), label='N(0, 2)', fill=True)
sns.kdeplot(normal(0, 3), label='N(0, 3)', fill=True)  
sns.kdeplot(normal(5, 1), label='N(0, 4)', fill=True)
sns.kdeplot(normal(-5, 2), label='N(0, 5)', fill=True)
plt.legend()
plt.title('KDE of Normal Distributions')
plt.xlabel('Value')
```

```python
# CDF of Normal Distributions
plt.figure(figsize=(10, 6))
sns.ecdfplot(normal(0, 1), label='N(0, 1)')
sns.ecdfplot(normal(0, 2), label='N(0, 2)')
sns.ecdfplot(normal(0, 3), label='N(0, 3)')
sns.ecdfplot(normal(5, 1), label='N(5, 1)')
sns.ecdfplot(normal(-5, 2), label='N(-5, 2)')
plt.legend()   
plt.title('CDF of Normal Distributions')
plt.xlabel('Value')
plt.show()
```

#### Standard Normal Variate - A Standard Normal Variate(Z) is a standardized form of the normal distribution with     mean = 0 and standard deviation = 1.

##### Example - Suppose weights of a students of a class follow Normal distribution with mean = 70 and std = 4

```python
from scipy.stats import norm
```

```python
x = 75
mean = 70
std = 4

# Probability of a value being less than or equal to x
probability_less_than_x = norm.cdf(x, loc=mean, scale=std)
print(f"Probability of value <= {x}: {probability_less_than_x}")

# Probability of a value being greater than x
probability_greater_than_x = 1 - norm.cdf(x, loc=mean, scale=std)
print(f"Probability of value > {x}: {probability_greater_than_x}")

# Probability of a value being between two points (a and b)
a = mean - 2 * std
b = mean + 2 * std
probability_between_a_b = np.abs(norm.cdf(b, loc=mean, scale=std) - norm.cdf(a, loc=mean, scale=std))
print(f"Probability of value between {a} and {b}: {probability_between_a_b}")

```

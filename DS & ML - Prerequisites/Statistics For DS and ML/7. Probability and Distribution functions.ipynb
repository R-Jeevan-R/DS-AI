#### Probability Distribution - It is a list of all possible values of a Random Variable along with their probability of occurence.
For example: In below exp. i.e summation of outcomes of rolling two dice, we can observe Random Variable taking values from 1 to 12 along with their probability.

```python
s3 = []
for _ in range(1000):
    a = np.random.randint(1,7)
    b = np.random.randint(1,7)
    s3.append(a + b)
s3 = pd.Series(s3)
```

```python
sns.displot(s3,kind='hist',kde=True,bins=12)
```

```python
s3.value_counts().sort_values()
```

```python
s4 = s3.value_counts() / s3.value_counts().sum()
```

#### Probability Distribution Function - A probability distribution function (PDF) is a mathematical function that describes the probability of obtaining different values of a random variable in a particular probability distribution.
2 types of Distribution functions:
1. Probability Mass Function (PMF) - For Discrete Random Variables
2. Probability Density Function (PDF) - For Continuous Random Variables

```python
sns.catplot(s4,kind='bar')
plt.title("Probability Distribution Function")
plt.xlabel("RV(X) - Sum of outcomes of rolling 2 dice")
plt.ylabel("Probability of RV(X)")
```

##### Probability Mass Functions - PMF stands for Probability Mass Function. It is a mathematical function that describes the probability distribution of a discrete random variable.
The PMF of a discrete random variable assigns a probability to each possible value
of the random variable. The probabilities assigned by the PMF must satisfy two
conditions:
1. The probability assigned to each value must be non-negative (i.e., greater
than or equal to zero).
2. The sum of the probabilities assigned to all possible values must equal 1.

##### 1 .Binomial Distribution - Binomial Distribution is a probability distribution that models the number of successes in a fixed number of independent, identical trials, where each trial can result in one of two outcomes, success or failure. 
PMF of Binomial : P(X = x) = COMB(n,x) * p^x * (1 - p)^(n-x)

```python
import math
p = 1/4
q = 1 - p

n = 10
s5 = pd.Series()

for i in range(1, n+1):
    s5[i] = math.comb(n,i)*(p**i)*(q**(n - i))
s5.plot(kind='bar')
```

```python
p = q = 0.5
n = 10
s6 = pd.Series()

for i in range(1, n+1):
    s6[i] = math.comb(n,i)*(p**i)*(q**(n - i))
fig,(a,b) = plt.subplots(1,2,figsize = (10,3))
a.bar(range(1,11),s6)
a.set_title("PMF of Binomail RV")
a.set_xlabel("RV(X)-Number of MCQs correct out of 10")
a.set_ylabel("Prob(X)")
b.plot(np.cumsum(s6))
b.set_title("CDF of Binomial RV")
b.set_xlabel("RV(X)-Number of MCQs correct out of 10")
```

##### 2. Bernoulli Distribution - Bernoulli Distribution includes a random variable with only two possible outcomes. When the variable is equal to 1, it represents success with the probability p. Whereas, when the variable is equal to 0, it shows failure. The probability of failure is q, which is calculated as 1-p.

```python
import numpy as np
import matplotlib.pyplot as plt
import math

np.random.seed(42)

dimensions = [2, 3, 10, 100, 1000, 10000]
num_samples = 1000

plt.figure(figsize=(14, 10))

for i, d in enumerate(dimensions, 1):
    # Generate random points in [0,1]^d
    v1 = np.random.rand(num_samples, d)
    v2 = np.random.rand(num_samples, d)

    # Euclidean distances
    distances = np.linalg.norm(v1 - v2, axis=1)

    plt.subplot(3, 2, i)
    plt.hist(distances, bins=100, density=True)
    plt.xlabel("Distance")
    plt.ylabel("Density")
    plt.title(f"{d} Dimensions")
    plt.xlim(0, math.sqrt(d))

plt.tight_layout()
plt.show()
```
